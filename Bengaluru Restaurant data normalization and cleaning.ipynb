# Cuisine table

```python
df['cuisines'].fillna('Unknown', inplace = True)
```

```python
df['cuisines'].isnull().sum()
```

```python
cuisine_series = df['cuisines'].str.split(',')
all_cuisines = set()
for sublist in cuisine_series:
    if isinstance(sublist, list):
        for cuisine in sublist:
            all_cuisines.add(cuisine.strip())
cuisine_df = pd.DataFrame(list(all_cuisines), columns=['cuisines']).reset_index(drop=True)
cuisine_df['cuisine_id'] = cuisine_df.index + 1
cuisine_df = cuisine_df[['cuisine_id', 'cuisines']]

cuisine_df.head()
```

# Restaurant table

```python
df['Unnamed: 0'].duplicated().sum()
```

```python
df.rename(columns = {'Unnamed: 0': 'restaurant_id'}, inplace = True)
```

```python
# Merge location & restaurant type IDs
restaurant_df = df[['restaurant_id', 'name', 'address', 'phone', 'url', 'location', 'rest_type', 'listed_in(type)', 'listed_in(city)', 'approx_cost(for two people)']].copy()
restaurant_df = restaurant_df.merge(location_df, on='location', how='left')
restaurant_df = restaurant_df.merge(rest_type_df, on='rest_type', how='left')
restaurant_df = restaurant_df.merge(listing_type_df, on='listed_in(type)', how='left')
restaurant_df = restaurant_df.merge(listing_city_df, on='listed_in(city)', how='left')
restaurant_df_final = restaurant_df[['restaurant_id', 'name', 'address', 'phone', 'url', 'location_id', 'rest_type_id', 'listing_id', 'listing_city_id', 'approx_cost(for two people)']]
restaurant_df_final.rename(columns={'approx_cost(for two people)': 'cost_for_two'}, inplace=True)
restaurant_df_final = restaurant_df_final[['restaurant_id', 'location_id', 'rest_type_id', 'listing_id', 'listing_city_id', 'name', 'address', 'phone', 'url', 'cost_for_two']]

restaurant_df_final.head()
```

# Restaurant-Cuisine Mapping Table

```python
restaurant_cuisine_rows = []
for idx, row in df[['restaurant_id', 'cuisines']].iterrows():
    cuisines = row['cuisines'].split(', ')
    for cuisine in cuisines:
        cuisine_id = cuisine_df.loc[cuisine_df['cuisines'] == cuisine.strip(), 'cuisine_id'].values[0]
        restaurant_cuisine_rows.append({'restaurant_id': row['restaurant_id'], 'cuisine_id': cuisine_id})
restaurant_cuisine_df = pd.DataFrame(restaurant_cuisine_rows)

restaurant_cuisine_df.head()
```
